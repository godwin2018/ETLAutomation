Group By - Using aggregate functions (sum, min,max, count,avg)

```python
# Using aggregate functions (sum, min,max, count,avg)
# Count the employees in each department
# Output should be the deptno, count_employees - expected output
# SQL - select deptno, count(*) from emp(df) group by deptno

df.groupby('deptno')['deptno'].count()
```

```python
aggr_data = df.groupby('deptno')['eno'].count()
type(aggr_data)
```

```python
# convert the Series to Dataframe
aggr_data = aggr_data.reset_index()
```

```python
aggr_data
```

```python
type(aggr_data)
```

```python
# rename eno colunm to count_employees
aggr_data.rename(columns={'eno':'count_employees'}, inplace=True)
```

```python
aggr_data
```

```python
# get total count of employee, max salary, min salary, total salary in each department
# SQL: select count(eno) as employees_count, max(salary) as min_salary, sum (salar) as total_salary group by deptno

df.groupby('deptno')['salary'].count()
```

```python
df.groupby('eno')['salary'].count().reset_index()
```

```python
agg_data = df.groupby('deptno')['salary'].agg(['count','max','min','sum'])
```

```python
agg_data
 #deptno behaves like index colunm
```

```python
type(agg_data)
```

```python
# to reset index
agg_data = agg_data.reset_index()
```

```python
agg_data
```

```python
# t rename colunm
agg_data.rename(columns={'salary':'count_employees','count':'employee_count','max':'max_salary','min':'min_salary','sum':'total_salary'}, inplace=True)
```

```python
agg_data
```

```python
agg_data.to_csv('emp_summary.csv',index=False)
```

Get total salary in each department except for location = Mumbai

```python
loc_filter = df['Location'] !='Mumbai'
loc_filter
```

```python
df_filtered = df[loc_filter]
df_filtered
```

```python
df_agg = df_filtered.groupby('deptno')['salary'].sum()
df_agg
```

```python
df_agg1 = df_agg.reset_index()
df_agg1
```

```python
df_agg1['salary'] > 50000
```

```python
# way 1
having_filter = df_agg1['salary'] > 50000
df_agg1[having_filter]
```

```python
# Way 2 - Using lambda
lambda_fil = df_agg.apply(lambda x : x > 50000)
df_agg[lambda_fil].reset_index()
```
