```python
import pandas as pd
```

```python
df_emp = pd.read_csv('data/employees.csv')
```

```python
df_emp
```

```python
df_dept = pd.read_csv('data/department.csv')
```

```python
df_dept
```

```python
df_sal = pd.read_csv('data/salary.csv')
```

```python
df_sal
```

Inner Join

```python
# SQL = slect * from employee as emp inner join departmen as dept on emp.deptno = dept.deptid
df_emp.merge(df_dept,left_on='deptno',right_on='deptid', how='inner')
```

```python
# SQL = slect * from employee as emp left join departmen as dept on emp.deptno = dept.deptid
df_emp.merge(df_dept,left_on='deptno',right_on='deptid', how='left')
```

```python
# SQL = select * from employee as emp right join departmen as dept on emp.deptno = dept.deptid
df_emp.merge(df_dept,left_on='deptno',right_on='deptid', how='right')
```

```python
# This can be achieved by using the left join and changing the dataframe sequence
df_dept.merge(df_emp,left_on='deptid',right_on='deptno', how='left')
```

```python
# full join
df_outer = df_emp.merge(df_dept,left_on='deptno',right_on='deptid', how='outer')
```

```python
df_outer
```

# Concat

```python
pd.concat([df_emp,df_dept],axis=0)
```

```python
pd.concat([df_emp,df_dept],axis=1)
```

```python
# Change the sequence
pd.concat([df_dept,df_emp],axis=1)
```

```python
# combine three dataframes
pd.concat([df_emp,df_dept,df_sal],axis=1)
```

```python
pd.concat([df_emp,df_dept,df_sal],axis=0)
```

```python
# Using join and concat
pd.concat([df_emp,df_dept], axis=1,join='inner')
```
